```python
from pydantic_ai import Agent
from pydantic_ai.mcp import MCPServerStreamableHTTP
import os , sys
# Pobierz katalog główny bam wzgledem obecnego folderu roboczego
bam_dir = os.path.abspath(os.path.join(os.getcwd(), ".."))
sys.path.insert(0, bam_dir)
from config import get_openrouter_model



server = MCPServerStreamableHTTP('https://mcp.alphavantage.co/mcp?apikey=IVHOSNJMC6')  
agent = Agent(get_openrouter_model(), toolsets=[server])  

async def main():
  result = await agent.run('Pobierz dzienne notowania Apple (AAPL) za ostatnie 10 dni używając funkcji TIME_SERIES_DAILY')
  print(result.output)
  #> The answer is 12.


await main()
```
